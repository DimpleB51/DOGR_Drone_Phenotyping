```python
from collections import OrderedDict
from datetime import datetime, time

# Helper: handle multiple datetime formats
def parse_datetime_safe(date_str):
    for fmt in ("%Y-%m-%d %H:%M:%S", "%d-%m-%Y %H:%M"):
        try:
            return datetime.strptime(date_str, fmt)
        except ValueError:
            continue
    raise ValueError(f"Date format not recognized: {date_str}")

# Define time slots
time_slots = OrderedDict({
    '4-8am':   (time(4, 0), time(8, 0)),
    '8-10am':  (time(8, 0), time(10, 0)),
    '10-12pm': (time(10, 0), time(12, 0)),
    '12-2pm':  (time(12, 0), time(14, 0)),
    '2-4pm':   (time(14, 0), time(16, 0))
})

# Step 1: Add 'date_only' and 'time_only' to each weather entry
for entry in ordered_dict_weather:
    dt = parse_datetime_safe(str(entry['Date(IST)']))
    entry['date_only'] = dt.date()
    entry['time_only'] = dt.time()

# Step 2: Create the nested OrderedDict
filtered_weather_by_trial = OrderedDict()

# Step 3: Populate with time-slot-grouped entries
for trial_name, date_list in updated_unique_dates.items():
    filtered_weather_by_trial[trial_name] = OrderedDict()
    for date_str in date_list:
        target_date = datetime.strptime(date_str, "%d/%m/%Y").date()
        # Initialize slot container
        slot_entries = OrderedDict((slot_name, []) for slot_name in time_slots)

        # Filter and assign to slot
        for entry in ordered_dict_weather:
            if entry['date_only'] == target_date:
                entry_time = entry['time_only']
                for slot_name, (start, end) in time_slots.items():
                    if start <= entry_time < end:
                        cleaned_entry = entry.copy()
                        cleaned_entry.pop('date_only', None)
                        cleaned_entry.pop('time_only', None)
                        slot_entries[slot_name].append(cleaned_entry)
                        break

        filtered_weather_by_trial[trial_name][date_str] = slot_entries

# Print result
from pprint import pprint
pprint(filtered_weather_by_trial['Trial7']['20/12/2023'])
```
