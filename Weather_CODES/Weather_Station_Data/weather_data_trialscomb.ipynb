```python
# Generating ordered_dict_weather 
import csv
from collections import OrderedDict
from datetime import datetime
import pandas as pd

# Initialize ordered_dict_weather as an empty list
ordered_dict_weather = []

# Load the CSV data and create the ordered_dict_weather list (Code 1)
with open('weather.csv', mode='r', newline='') as file:
    reader = csv.DictReader(file)
    for row in reader:
        ordered_row = OrderedDict()
        for key, value in row.items():
            if key == 'Date(IST)':  # Modify this if your column name differs
                # Try to standardize the date format if necessary
                value = value.strip()
                # Replace '/' or '.' with '-' to handle common formats
                value = value.replace('/', '-').replace('.', '-')
                # Now parse it
                try:
                    dt = datetime.strptime(value, "%d-%m-%Y %H:%M")
                except ValueError:
                    try:
                        dt = datetime.strptime(value, "%d-%m-%Y")
                    except ValueError:
                        dt = None  # Or handle it as needed
                value = dt
            ordered_row[key] = value
        ordered_dict_weather.append(ordered_row)

# Sort by the parsed datetime object
DATE_COLUMN = 'Date(IST)'  # Change if needed
ordered_dict_weather = [row for row in ordered_dict_weather if row[DATE_COLUMN] is not None]
ordered_dict_weather.sort(key=lambda x: x[DATE_COLUMN])

# Load the Excel data and create the records_filtered list (Code 2)
# Load all sheets from the Excel file
xls = pd.read_excel('Jan-Sept_2024_Weather_data.xlsx', sheet_name=None)

# Get specific sheet (datewise-data)
df_weather24 = xls.get('datewise-data')

# Convert 'Date & Time' to datetime format
df_weather24['Date & Time'] = pd.to_datetime(df_weather24['Date & Time'], errors='coerce')

# Filter data between January and April
df_filtered = df_weather24[(df_weather24['Date & Time'].dt.month >= 1) & (df_weather24['Date & Time'].dt.month <= 4)]

# Select the relevant columns
columns_to_keep = ['Date & Time', 'Air Temperature (Deg C)', 'Air RH (%)', 'Rainfall (mm)']
df_filtered = df_filtered[columns_to_keep]

# Create an OrderedDict with the new keys
records_filtered = []
for _, row in df_filtered.iterrows():
    # Format the Date & Time to 'DD-MM-YYYY HH:MM' (IST)
    formatted_date = row['Date & Time'].strftime('%d-%m-%Y %H:%M')
    
    # Create OrderedDict with new keys and formatted values
    row_dict = OrderedDict({
        'Date(IST)': formatted_date,
        'Temp': f"{row['Air Temperature (Deg C)']:.2f}",
        'RH(%)': f"{row['Air RH (%)']:.2f}",
        'Rain(mm)': f"{row['Rainfall (mm)']:.2f}"
    })
    records_filtered.append(row_dict)

# Append records_filtered to ordered_dict_weather
ordered_dict_weather += records_filtered  # Or you can use ordered_dict_weather.extend(records_filtered)

# Print the total number of rows in the combined list
print(f"Total rows after appending: {len(ordered_dict_weather)}")

# Optional: Convert datetime back to string if printing
for item in ordered_dict_weather[43810:43832]:  # Print the first 10 records
    dt_obj = item[DATE_COLUMN]
    item[DATE_COLUMN] = dt_obj.strftime("%d-%m-%Y %H:%M") if isinstance(dt_obj, datetime) else dt_obj
    print(item)
```

```python
from collections import OrderedDict
import numpy as np

weather_master = OrderedDict()

for trial_key in filtered_weather_by_trial:
    weather_master[trial_key] = OrderedDict()

    for date_curr in filtered_weather_by_trial[trial_key]:
        weather_master[trial_key][date_curr] = OrderedDict()

        deltaRH = OrderedDict()
        temp_curr_day = OrderedDict()
        RH_curr_day = OrderedDict()
        rain_curr_day = OrderedDict()

        for time_block in filtered_by_past7days[trial_key][date_curr]:
            temp_vals = []
            RH_vals = []
            rain_vals = []

            for data in filtered_weather_by_trial[trial_key][date_curr][time_block]:
                temp_vals.append(float(data['Temp']))
                RH_vals.append(float(data['RH(%)']))
                rain_vals.append(float(data['Rain(mm)']))

            # Store time-block-wise means to compute daily aggregates
            deltaRH[time_block] = np.mean(RH_vals)
            temp_curr_day[time_block] = np.mean(temp_vals)
            RH_curr_day[time_block] = np.mean(RH_vals)
            rain_curr_day[time_block] = np.mean(rain_vals)

        # Final daily metrics stored without including raw time blocks
        try:
            RHdelta_value = deltaRH['4-8am'] - np.mean([deltaRH['10-12pm'], deltaRH['12-2pm']])
        except KeyError:
            RHdelta_value = None  # handle missing blocks gracefully

        weather_master[trial_key][date_curr]['RHdelta'] = RHdelta_value
        weather_master[trial_key][date_curr]['temp_curr_day'] = np.mean(list(temp_curr_day.values()))
        weather_master[trial_key][date_curr]['RH_curr_day'] = np.mean(list(RH_curr_day.values()))
        weather_master[trial_key][date_curr]['rain_curr_day'] = np.mean(list(rain_curr_day.values()))
        weather_master[trial_key][date_curr]['rain_total_day'] = sum(rain_curr_day.values())

# Print a sample to verify
for key, value in weather_master['Trial1'].items():
    print(key, value)
    print("\n")
```

```python
from collections import OrderedDict, defaultdict
import numpy as np
import pandas as pd

# Step 1: Aggregate date-wise metrics from weather_master
temp_storage = defaultdict(lambda: defaultdict(list))

for trial_key in weather_master:
    for date_key, metrics in weather_master[trial_key].items():
        for metric_key, value in metrics.items():
            if value is not None:
                temp_storage[date_key][metric_key].append(value)

date_wise_weather = OrderedDict()

for date_key, metric_groups in temp_storage.items():
    date_wise_weather[date_key] = {}
    for metric_key, value_list in metric_groups.items():
        date_wise_weather[date_key][metric_key] = float(np.mean(value_list))

# Step 2: Convert to DataFrame
df = pd.DataFrame.from_dict(date_wise_weather, orient='index')
df.index.name = 'Date'
df.reset_index(inplace=True)

# Step 3: Export to Excel
output_file = 'aggregated_weather_data.xlsx'
df.to_excel(output_file, index=False)

print(f"\n✅ Exported successfully to {output_file}")
```
