```python
# Master excel sheet
master = OrderedDict()

for trial_key in ordered_data_by_sheet:
    trial_data = ordered_data_by_sheet[trial_key]
    if trial_key not in master:
        master[trial_key] = OrderedDict()

    for row in trial_data:
        if row.get('Treatment') != 'T4':
            continue

        date = row.get('Date')
        plot_name = row.get('Name of plot')

        if not date or not plot_name:
            continue

        if date not in master[trial_key]:
            master[trial_key][date] = OrderedDict()

        # Build the required output values
        values = OrderedDict({
            'DAT': row.get('DAT'),
            'PDI_SB_new': row.get('PDI_SB_new'),
            'new_thrips': row.get('new_thrips'),
            'PDI_PB_new': row.get('PDI_PB_new'),
            'PDI_AN_new': row.get('PDI_AN_new')
        })

        master[trial_key][date][plot_name] = values

print (master['Trial1'])
#print ("\n\n")
#print (master['Trial5']['20/10/2023'])
```

```python
from collections import defaultdict, OrderedDict
import re

def group_data(entry_dict):
    grouped = defaultdict(OrderedDict)
    
    for key, value in entry_dict.items():
        # Skip 'Date' only — but **keep** RHdelta
        if key.lower() == 'date':
            continue

        # Group by key suffixes
        if '_day' in key:
            grouped['current_day'][key] = value
        elif '_prevday' in key:
            grouped['previous_day'][key] = value
        elif '_last_7days' in key:
            grouped['last_7_days'][key] = value
        else:
            grouped['others'][key] = value
    
    return grouped

# Main loop: group all trials and dates
grouped_weather = OrderedDict()

for trial_name, trial_data in weather_eda.items():
    grouped_weather[trial_name] = OrderedDict()
    for date, daily_data in trial_data.items():
        grouped_weather[trial_name][date] = group_data(daily_data)

# Example: pretty print for a specific trial and date
from pprint import pprint
pprint(grouped_weather['Trial9']['30/12/2023'])
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

flat_data = []

for trial_name, trial_data in split_region_data.items():
    for region, region_data in trial_data.items():
        for block_name, entries in region_data.items():
            for entry in entries:
                flat_entry = {
                    'Trial': trial_name,
                    'Region': region,
                    'Block': block_name,
                    'Date': entry['Date']
                }

                # Include all other keys (metrics + weather)
                for key, val in entry.items():
                    if key != 'Date':
                        flat_entry[key] = val

                flat_data.append(flat_entry)

# Create DataFrame
df = pd.DataFrame(flat_data)

# Convert all columns to numeric where possible
df = df.apply(pd.to_numeric, errors='ignore')

# Choose one metric for correlation
target_metric = 'PDI_SB_new'

# Remove rows where target metric is missing, inf, or nan
df = df[np.isfinite(df[target_metric])]

# Compute correlation of the target metric with each weather parameter
print(f"Correlation of '{target_metric}' with weather parameters:\n")

for col in df.columns:
    if col in ['Trial', 'Region', 'Block', 'Date', target_metric]:
        continue

    if pd.api.types.is_numeric_dtype(df[col]):
        # Clean this column too
        valid_idx = np.isfinite(df[col])
        x = df[target_metric][valid_idx]
        y = df[col][valid_idx]

        if len(x) > 2:
            corr, pval = pearsonr(x, y)
            print(f"{col:15s}: Correlation = {corr:.3f}, p-value = {pval:.4f}")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# -----------------------------
# Step 1: Flatten split_region_data
# -----------------------------
flat_data = []

for trial_name, trial_data in split_region_data.items():
    for region, region_data in trial_data.items():
        for block_name, entries in region_data.items():
            for entry in entries:
                flat_entry = {
                    'Trial': trial_name,
                    'Region': region,
                    'Block': block_name,
                    'Date': entry['Date']
                }
                for key, val in entry.items():
                    if key != 'Date':
                        flat_entry[key] = val
                flat_data.append(flat_entry)

df = pd.DataFrame(flat_data)

# -----------------------------
# Step 2: Clean and filter data
# -----------------------------

# Convert all columns to numeric where possible
df = df.apply(pd.to_numeric, errors='ignore')

# Remove non-numeric and constant columns
numeric_df = df.select_dtypes(include=[np.number])
numeric_df = numeric_df.loc[:, numeric_df.std() != 0]

# -----------------------------
# Step 3: Loop through all metrics
# -----------------------------

# Define the list of disease metrics you want to analyze
target_metrics = ['PDI_SB_new', 'new_thrips', 'PDI_PB_new', 'PDI_AN_new']

# Optional: store results
results = []

for metric in target_metrics:
    if metric not in numeric_df.columns:
        continue

    print(f"\n--- Correlation for: {metric} ---")
    for col in numeric_df.columns:
        if col == metric:
            continue

        # Drop NaNs and Infs from both variables
        x = numeric_df[metric].replace([np.inf, -np.inf], np.nan)
        y = numeric_df[col].replace([np.inf, -np.inf], np.nan)
        valid = x.notna() & y.notna()

        if valid.sum() > 2:
            try:
                corr, pval = pearsonr(x[valid], y[valid])
                print(f"{col:25s} : Correlation = {corr:+.3f}, p-value = {pval:.4f}")
                results.append({
                    'Metric': metric,
                    'Variable': col,
                    'Correlation': round(corr, 4),
                    'P-value': round(pval, 6)
                })
            except Exception as e:
                print(f"{col:25s} : Error - {e}")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
import seaborn as sns
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Flatten split_region_data
# -----------------------------
flat_data = []

for trial_name, trial_data in split_region_data.items():
    for region, region_data in trial_data.items():
        for block_name, entries in region_data.items():
            for entry in entries:
                flat_entry = {
                    'Trial': trial_name,
                    'Region': region,
                    'Block': block_name,
                }
                for key, val in entry.items():
                    if key != 'Date':  # Exclude 'Date' from features
                        flat_entry[key] = val
                flat_data.append(flat_entry)

df = pd.DataFrame(flat_data)

# -----------------------------
# Step 2: Clean numeric data
# -----------------------------
df = df.apply(pd.to_numeric, errors='ignore')
numeric_df = df.select_dtypes(include=[np.number])
numeric_df = numeric_df.loc[:, numeric_df.std() != 0]  # Remove constant columns

# -----------------------------
# Step 3: Build Correlation Matrix
# -----------------------------
# Define metrics and infer weather columns
target_metrics = ['PDI_SB_new', 'new_thrips', 'PDI_PB_new', 'PDI_AN_new']
weather_vars = [col for col in numeric_df.columns if col not in target_metrics]

# Create correlation matrix (weather vars = rows, metrics = columns)
correlation_matrix = pd.DataFrame(index=weather_vars, columns=target_metrics, dtype=float)

for metric in target_metrics:
    for var in weather_vars:
        x = numeric_df[metric].replace([np.inf, -np.inf], np.nan)
        y = numeric_df[var].replace([np.inf, -np.inf], np.nan)
        valid = x.notna() & y.notna()

        if valid.sum() > 2:
            corr, _ = pearsonr(x[valid], y[valid])
            correlation_matrix.loc[var, metric] = corr

# -----------------------------
# Step 4: Plot Heatmap (Metrics on X-axis)
# -----------------------------
plt.figure(figsize=(12, len(weather_vars) * 0.4))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Weather Parameters vs. Disease Metrics Correlation", fontsize=14)
plt.ylabel("Weather Parameters")
plt.xlabel("Disease Metrics")
plt.xticks(rotation=0)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
