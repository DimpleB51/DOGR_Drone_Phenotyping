```python
import csv
from collections import OrderedDict
from datetime import datetime

ordered_dict_sub = []

with open('pdi_data.csv', mode='r', newline='') as file:
    reader = csv.DictReader(file)
    for row in reader:
        ordered_row = OrderedDict()
        for key, value in row.items():
            # Check if value looks like a date in dd.mm.yyyy format
            if isinstance(value, str) and '.' in value and len(value) >= 8:
                parts = value.split('.')
                if len(parts) == 3 and all(part.isdigit() for part in parts):
                    day = parts[0].zfill(2)
                    month = parts[1].zfill(2)
                    year = parts[2]
                    value = f"{day}/{month}/{year}"  # Format: dd/mm/yyyy
            ordered_row[key] = value
        ordered_dict_sub.append(ordered_row)

# Sort by date column
DATE_COLUMN = 'Date'  # Change if your column header is different

ordered_dict_sub.sort(key=lambda x: datetime.strptime(x[DATE_COLUMN], "%d/%m/%Y"))

# Output

print(f"Total rows: {len(ordered_dict_sub)}")
for item in ordered_dict_sub:
    print(item)
```

```python
from copy import deepcopy

updated_main_sheet = OrderedDict()
for trial_key in ordered_data_by_sheet:
    if 5<= int(trial_key[-1]) <=9: # restrict trials between 5 and 9
        print (trial_key)
        if trial_key in ordered_data_by_sheet:
            trial_data = ordered_data_by_sheet[trial_key]
            # Prepare to hold enriched rows
            if trial_key not in updated_main_sheet:
                updated_main_sheet[trial_key] = []
            #for main_row in trial_data[:1]:
            for main_row in trial_data:
                date_found_flag = False
                dates_in_trials = []
                for sub_row in ordered_dict_sub:
                    #print(sub_row)
                    sub_trial_key = sub_row.get('\ufeffTrial name') or sub_row.get('Trial name')
                    sub_trial_key = sub_trial_key.replace(" ", "")
                    #print (sub_trial_key)
                    
                    updated_row = deepcopy(main_row)
                    # Match condition
                    if (
                        trial_key == sub_trial_key and
                        str(main_row.get('Treatment')) == str(sub_row.get('Treatment')) and
                        str(main_row.get('Replication')) == str(sub_row.get('Replication')) and
                        str(main_row.get('Date')) == str(sub_row.get('Date'))
                    ):
                     
                        # Step 7: Add new values from row
                        updated_row['PDI_SB_new'] = str(sub_row.get('PDI_SB'))
                        updated_row['PDI_PB_new'] = str(sub_row.get('PDI_PB'))
                        updated_row['PDI_AN_new'] = str(sub_row.get('PDI_AN'))            
                        # Step 8: Add to final dictionary
                        
                        updated_main_sheet[trial_key].append(updated_row)
                        date_found_flag = True
                    else:
                        dates_in_trials.append(str(sub_row.get('Date')))
                if date_found_flag == False:
                    main_row_date = str(main_row.get('Date'))
                    main_row_date = datetime.strptime(main_row_date, "%d/%m/%Y")
                    subrows = list(set(dates_in_trials))
                    # Convert to datetime objects
                    subrow_dates = [datetime.strptime(date_str, "%d/%m/%Y") for date_str in subrows]
                    
                    # Find the closest date
                    closest_date = min(subrow_dates, key=lambda d: abs(d - main_row_date))
                    closest_date = closest_date.strftime("%d/%m/%Y")
                    #print ("main_row_date: ", main_row_date, "closest_date: ", closest_date)
                    for sub_row in ordered_dict_sub:
                        sub_trial_key = sub_row.get('\ufeffTrial name') or sub_row.get('Trial name')
                        sub_trial_key = sub_trial_key.replace(" ", "")
                        updated_row = deepcopy(main_row)
                        # Match condition
                        if (
                            trial_key == sub_trial_key and
                            closest_date == str(sub_row.get('Date')) and 
                            str(main_row.get('Treatment')) == str(sub_row.get('Treatment')) and
                            str(main_row.get('Replication')) == str(sub_row.get('Replication'))                            
                        ):
                            # Step 7: Add new values from row
                            updated_row['PDI_SB_new'] = str(sub_row.get('PDI_SB'))
                            updated_row['PDI_PB_new'] = str(sub_row.get('PDI_PB'))
                            updated_row['PDI_AN_new'] = str(sub_row.get('PDI_AN'))            
                            # Step 8: Add to final dictionary
                            
                            updated_main_sheet[trial_key].append(updated_row)
```

```python
import pandas as pd

# Save to a new Excel file
output_filename = "updated_trials.xlsx"

with pd.ExcelWriter(output_filename, engine='xlsxwriter') as writer:
    print(f"Total sheets: {len(updated_main_sheet)}")
    
    for sheet_name, row_list in updated_main_sheet.items():
        print(f"Writing sheet: {sheet_name} with {len(row_list)} rows")

        # Convert list of dicts to DataFrame
        df = pd.DataFrame(row_list)

        # Excel sheet names must be <=31 characters and cannot contain certain characters
        safe_sheet_name = sheet_name[:31].replace('/', '_').replace('\\', '_').replace('*', '_').replace('[', '_').replace(']', '_')

        # Write to sheet
        df.to_excel(writer, sheet_name=safe_sheet_name, index=False)
```
