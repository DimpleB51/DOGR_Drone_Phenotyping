```python
import pandas as pd
from collections import OrderedDict
from pathlib import Path

# === Inputs: replace these with your actual variables if different names ===
ordered_data_by_sheet = ordered_data_by_sheet   # your dict of sheets -> list(OrderedDict) (sample provided)
weather_eda = weather_eda                       # your dict of sheet -> OrderedDict(date -> OrderedDict)

# Output file
out_path = Path("merged_with_weather.xlsx")

def to_dataframe_from_rows(rows):
    """Convert rows which may be: list of OrderedDicts, OrderedDict of rows, or dict -> DataFrame"""
    if rows is None:
        return pd.DataFrame()
    # If it's a mapping of row-id -> row-dict
    if isinstance(rows, (dict, OrderedDict)):
        # if values are the row dicts
        vals = list(rows.values())
        # if vals themselves are scalars (unlikely), turn to dict
        if vals and not isinstance(vals[0], (dict, OrderedDict)):
            # maybe rows is like {1: OrderedDict(...), 2: ...} handled above, else fallback
            return pd.DataFrame([dict(rows)])
        return pd.DataFrame([dict(v) for v in vals])
    # If it's an iterable/list of OrderedDict
    try:
        return pd.DataFrame([dict(r) for r in rows])
    except Exception:
        return pd.DataFrame(rows)

def to_dataframe_from_weather(wdict):
    """Convert weather_eda[sheet] (mapping date -> OrderedDict) into a DataFrame.
       Ensures 'Date' column exists."""
    if wdict is None:
        return pd.DataFrame()
    # wdict expected like OrderedDict({ '28/08/2023': OrderedDict({...}), ... })
    rows = []
    for key, val in wdict.items():
        # val may already include 'Date', if not set it from the key
        row = dict(val) if isinstance(val, (dict, OrderedDict)) else {}
        if 'Date' not in row or not row.get('Date'):
            row['Date'] = key
        rows.append(row)
    return pd.DataFrame(rows)

def normalize_date_col(df, colname='Date'):
    """Normalize a Date-like column to datetime.date (dayfirst=True) if present."""
    if colname in df.columns:
        df[colname] = pd.to_datetime(df[colname], dayfirst=True, errors='coerce').dt.date
    return df

# Create ExcelWriter and write merged sheets
with pd.ExcelWriter(out_path, engine='openpyxl') as writer:
    for sheet_name, rows in ordered_data_by_sheet.items():
        # Convert original data rows into DataFrame
        df_main = to_dataframe_from_rows(rows)
        if df_main.empty:
            print(f"Warning: sheet {sheet_name} produced empty DataFrame from ordered_data_by_sheet.")
        # Convert weather for this sheet (if any)
        weather_for_sheet = weather_eda.get(sheet_name) if weather_eda is not None else None
        df_weather = to_dataframe_from_weather(weather_for_sheet)

        # Normalize dates (both DataFrames)
        df_main = normalize_date_col(df_main, 'Date')
        df_weather = normalize_date_col(df_weather, 'Date')

        # If main Date column couldn't be parsed, try alternative column names (common mistakes)
        if 'Date' not in df_main.columns or df_main['Date'].isna().all():
            # try capitalizations or strip
            for alt in ['date', 'DATE']:
                if alt in df_main.columns:
                    df_main['Date'] = pd.to_datetime(df_main[alt], dayfirst=True, errors='coerce').dt.date
                    break

        # Merge (left join to keep all original rows). Weather columns will get appended.
        if not df_weather.empty and 'Date' in df_weather.columns:
            merged = df_main.merge(df_weather, on='Date', how='left', suffixes=('', '_weather'))
        else:
            merged = df_main.copy()
            # if weather exists but couldn't parse dates, try merging by string match
            if not df_weather.empty:
                # try merging on string dates (original string format)
                # rebuild weather df with original key as date_str
                df_weather_alt = pd.DataFrame([dict(v) for v in weather_for_sheet.values()])
                if 'Date' in df_weather_alt.columns:
                    merged = df_main.merge(
                        df_weather_alt,
                        left_on=df_main['Date'].astype(str),
                        right_on=df_weather_alt['Date'].astype(str),
                        how='left',
                        suffixes=('', '_weather')
                    ).drop(columns=['key_0'], errors='ignore')

        # Write merged sheet to Excel
        out_sheet_name = f"{sheet_name}_merged"
        # Excel sheet names must be <=31 chars
        out_sheet_name = out_sheet_name[:31]
        merged.to_excel(writer, sheet_name=out_sheet_name, index=False)

print(f"Saved merged workbook to: {out_path.resolve()}")
```

```python
import pandas as pd
from collections import OrderedDict
from pathlib import Path

ordered_data_by_sheet = ordered_data_by_sheet   # your variable in the notebook
weather_eda = weather_eda                       # your variable in the notebook

out_path = Path("merged_with_weather_resilient.xlsx")

def rows_to_df(rows):
    if rows is None:
        return pd.DataFrame()
    if isinstance(rows, (dict, OrderedDict)):
        vals = list(rows.values())
        if vals and isinstance(vals[0], (dict, OrderedDict)):
            return pd.DataFrame([dict(v) for v in vals])
        return pd.DataFrame([dict(rows)])
    return pd.DataFrame([dict(r) for r in rows])

def weather_to_df(wdict):
    if wdict is None:
        return pd.DataFrame()
    rows = []
    for key, val in wdict.items():
        row = dict(val) if isinstance(val, (dict, OrderedDict)) else {}
        row_date = row.get('Date') or key
        row['Date'] = row_date
        rows.append(row)
    return pd.DataFrame(rows)

def ensure_date_string(df, col_candidates=('Date','date','DATE')):
    if 'Date_str' in df.columns:
        return df
    for c in col_candidates:
        if c in df.columns and not df[c].isna().all():
            try:
                dt = pd.to_datetime(df[c], dayfirst=True, errors='coerce')
                df['Date_str'] = dt.dt.strftime('%d/%m/%Y')
                mask_na = df['Date_str'].isna()
                if mask_na.any():
                    df.loc[mask_na, 'Date_str'] = df.loc[mask_na, c].astype(str).str.strip()
                return df
            except Exception:
                df['Date_str'] = df[c].astype(str).str.strip()
                return df
    for c in df.columns:
        sample = df[c].astype(str).fillna('').iloc[0]
        if '/' in sample:
            df['Date_str'] = df[c].astype(str).str.strip()
            return df
    df['Date_str'] = pd.NA
    return df

written_any = False

with pd.ExcelWriter(out_path, engine='openpyxl') as writer:
    for sheet_name, rows in ordered_data_by_sheet.items():
        df_main = rows_to_df(rows)
        df_weather = weather_to_df(weather_eda.get(sheet_name) if weather_eda else None)

        df_main = ensure_date_string(df_main, col_candidates=('Date','date','DATE'))
        df_weather = ensure_date_string(df_weather, col_candidates=('Date','date','DATE'))

        df_main['Date_str'] = df_main['Date_str'].astype(str).str.strip()
        df_weather['Date_str'] = df_weather['Date_str'].astype(str).str.strip()

        if not df_weather.empty:
            df_weather_for_merge = df_weather.copy()
            merged = df_main.merge(df_weather_for_merge, on='Date_str', how='left', suffixes=('','_weather'))
        else:
            merged = df_main.copy()

        weather_cols = [c for c in df_weather.columns if c not in df_main.columns and c != 'Date_str']
        if weather_cols:
            merged['weather_matched'] = merged[weather_cols].notna().any(axis=1)
        else:
            merged['weather_matched'] = merged['Date_str'].isin(df_weather['Date_str']) if not df_weather.empty else False

        # move Date_str to front and rename to Date
        cols = list(merged.columns)
        if 'Date_str' in cols:
            cols = ['Date_str'] + [c for c in cols if c != 'Date_str']
            merged = merged[cols]
            merged = merged.rename(columns={'Date_str': 'Date'})

        # safe retrieval of sample date values even if multiple 'Date' columns exist
        sample_dates = "NO DATE"
        if 'Date' in merged.columns:
            # find indices of all columns named 'Date'
            date_indices = [i for i, c in enumerate(merged.columns) if c == 'Date']
            collected = []
            for idx in date_indices:
                col_series = merged.iloc[:, idx].astype(str).fillna('').head(5)
                # extend collected with values (avoid calling .tolist() on DataFrame)
                for v in col_series.values:
                    if v not in collected:
                        collected.append(v)
                    if len(collected) >= 5:
                        break
                if len(collected) >= 5:
                    break
            sample_dates = collected[:5] if collected else [""]  # list of strings

        # write sheet (ensure workbook has visible sheets)
        out_sheet_name = (sheet_name + "_merged")[:31]
        if merged.empty:
            pd.DataFrame({'Note': [f'No rows for sheet {sheet_name}']}).to_excel(writer, sheet_name=out_sheet_name, index=False)
        else:
            merged.to_excel(writer, sheet_name=out_sheet_name, index=False)

        # ensure visible
        try:
            wb = writer.book
            if out_sheet_name in wb.sheetnames:
                ws = wb[out_sheet_name]
                ws.sheet_state = 'visible'
        except Exception:
            pass

        written_any = True
        first_cols = list(merged.columns)[:10]
        print(f"Sheet '{sheet_name}': columns -> {first_cols}")
        print(f"  Sample Date values (up to 5 unique): {sample_dates}")
        print(f"  Weather columns detected (up to 10): {weather_cols[:10]}")

    if not written_any:
        pd.DataFrame({'Note': ['No sheets merged — check inputs']}).to_excel(writer, sheet_name='empty', index=False)
        try:
            writer.book['empty'].sheet_state = 'visible'
        except Exception:
            pass

print(f"Saved merged workbook to: {out_path.resolve()}")
```
