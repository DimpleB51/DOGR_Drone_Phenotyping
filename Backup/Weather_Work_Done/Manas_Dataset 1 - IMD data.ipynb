```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from datetime import datetime,timedelta

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_excel('D:\Shital\DOGR Thrips_data_2021_2023\IMD weather data and thrips data\MAHARASHTRA_PUNE_RAJGURUNAGAR_18.841_73.884_598.13_2018-01-01_2024-01-10_AWS.xlsx')
df
```

```python
df['DATE'] = pd.to_datetime(df['DATE'])
df['year'], df['month'] = df['DATE'].dt.year, df['DATE'].dt.month
df
```

```python
df_2022 = df[(df['year'] == 2022) ]
df_2022 = df_2022[(df_2022['month']== 12)]
df_2022 = df_2022[(df_2022['DATE']== '2022-12-31')]
df_2022
```

```python
df_2023 = df[(df['year'] == 2023) ]
df_2023
```

```python
df_2023 = pd.concat([df_2022,df_2023])
df_2023
```

```python
df_2023 = df_2023[['DATE','TIME(UTC)','RAIN_DAILY(mm)','TEMP(C)','TEMP_MIN(C)','TEMP_MAX(C)','RH(%)','year','month']]
df_2023.reset_index(inplace=True,drop=True)
df_2023
```

```python
df_2023['DATE'] = df_2023['DATE'].astype(str)
df_2023['TIME(UTC)'] = df_2023['TIME(UTC)'].astype(str)
df_2023['Date'] = (df_2023['DATE'] + ' ' + df_2023['TIME(UTC)'])

# Drop the original 'DATE' and 'TIME' columns if needed
# df_2023 = df_2023.drop(['DATE', 'TIME(UTC)'], axis=1)
column_order = ['Date'] + [col for col in df_2023.columns if col != 'Date']
df_2023 = df_2023[column_order]
df_2023
```

```python
df_2023['Date'] = pd.to_datetime(df_2023['Date'])
df_2023['Hour'] = df_2023['Date'].dt.hour
df_2023
```

```python
df_2023 = df_2023.groupby(df_2023['Date'].dt.floor('H')).first().reset_index(drop=True)
```

```python
df_2023.drop(columns=['DATE','TIME(UTC)','year','month','Hour'],inplace=True)
df_2023
```

```python
import matplotlib.pyplot as plt
from datetime import datetime, timedelta

# Create a figure and axis with shared x-axis and custom figure size
fig, axs = plt.subplots(1, 1, sharex=True, figsize=(14, 4))

# Define the start and end dates
start_date = datetime(2022, 12, 30)
end_date = datetime(2023, 3, 20)

# Define the date ranges and colors for axvspan
date_ranges = []
current_date = start_date
color = 'gray'
while current_date < end_date:
    next_date = current_date + timedelta(days=1)
    date_ranges.append((current_date, next_date, color))
    color = 'gray' if color == 'lightgray' else 'lightgray'
    current_date = next_date

# Scatter plot of data
axs.scatter(Rabi_2023['Date'], Rabi_2023['Thrips/Plant'], color='r', s=30)
axs.set_ylabel('Thrips/Plant', fontsize=9)  # Adjusted font size
axs.tick_params(axis='y', labelsize=9)      # Adjusted tick label size
axs.grid(axis='both')
axs.set_yticks(range(0, 101, 20))
axs.set_ylim(-2, 100)  # Set y-axis limits

axs.set_xlim(start_date, end_date)

# Set x-axis ticks and labels for weekly intervals
date_ticks = [start_date + timedelta(weeks=i) for i in range(((end_date - start_date).days + 7) // 7)]
date_labels = [date.strftime('%d-%b') for date in date_ticks]
axs.set_xticks(date_ticks)
axs.set_xticklabels(date_labels, fontsize=9)  # Adjusted tick label size
axs.tick_params(axis='x', labelsize=9)        # Adjusted tick label size

plt.subplots_adjust(hspace=0.9)
plt.suptitle('Rabi 2023(Dec22-Mar2023)', fontsize=12, fontweight='bold')  # Adjusted title font size
plt.tight_layout(rect=[0.1, 0.5,0.9, 1.05])

# Add the date ranges with alternating colors
for start, end, color in date_ranges:
    axs.axvspan(start, end, facecolor=color, alpha=0.18)

# Display the plot
plt.show()
```
